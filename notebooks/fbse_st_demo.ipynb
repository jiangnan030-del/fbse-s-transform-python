# FBSE-ST demo notebook

This notebook generates a synthetic non-stationary signal, runs the Fourier-Bessel domain S transform, and visualizes the result.

```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt

ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
SRC = ROOT / 'src'
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

from fbst.transform import fbse_s_transform
from fbst.visualization import plot_signal_and_time_frequency
```

```python
t = np.linspace(0.0, 1.0, 256, endpoint=False)
signal = (
    np.sin(2 * np.pi * (10 * t + 35 * t**2))
    + 0.6 * np.sin(2 * np.pi * 60 * t)
    + np.exp(-320 * (t - 0.68) ** 2) * np.sin(2 * np.pi * 88 * t)
)
result = fbse_s_transform(signal, num_zeros=256, sigma_scale=0.08)
result['time_frequency_matrix'].shape
```

```python
fig, axes = plot_signal_and_time_frequency(
    signal,
    result['energy_time_frequency_matrix'],
    title='FBSE-ST notebook demo',
)
plt.show()
```
