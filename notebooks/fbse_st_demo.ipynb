# FBSE-ST and STFT comparison notebook

This notebook generates a synthetic non-stationary signal, runs the Fourier-Bessel domain S transform, computes an STFT baseline, and visualizes the comparison.

```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import stft

ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
SRC = ROOT / 'src'
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

from fbst.transform import fbse_s_transform
from fbst.visualization import plot_fbse_stft_comparison
```

```python
t = np.linspace(0.0, 1.0, 256, endpoint=False)
signal = (
    np.sin(2 * np.pi * (8 * t + 36 * t**2))
    + 0.7 * np.sin(2 * np.pi * (20 * t + 10 * t**2))
    + 0.9 * np.exp(-350 * (t - 0.62) ** 2) * np.sin(2 * np.pi * 85 * t)
)
fbse_result = fbse_s_transform(signal, num_zeros=256, sigma_scale=0.08, toeplitz_mode='page_style', normalize_output=True)
_, _, zxx = stft(signal, window='hann', nperseg=64, noverlap=56, boundary=None)
stft_energy = np.abs(zxx) ** 2
fbse_result['energy_time_frequency_matrix'].shape, stft_energy.shape
```

```python
fig, axes = plot_fbse_stft_comparison(
    signal,
    fbse_result['energy_time_frequency_matrix'],
    stft_energy,
)
plt.show()
```
